```python
import yfinance as yf
import pandas as pd
D = yf.download('AAPL', period='2y')
prix = D['Close']
SMA20=prix.rolling(20).mean()
signal_brut=(prix > SMA20).astype(int) #prix > SMA20 compare jour par jour → donne une série de True/False /.astype(int) transforme ces True/False en 1/0
signal=signal_brut.shift(1)
print(SMA20.head(25))
print(SMA20.tail())
print(signal.tail())
```
