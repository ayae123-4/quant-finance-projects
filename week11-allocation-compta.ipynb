```python
from indicateurs import charger_prix, signal_SMA

prix=charger_prix('AAPL')
signal=signal_SMA(prix, 20)

cash = 10000
positions = 0
valeurs = []

for date in prix.index:
    prix_jour=prix[date]
    sig=signal[date]
    if sig==1 and positions==0:
        nb_actions=int(cash/prix_jour)
        cash=cash-nb_actions*prix_jour
        positions=positions+nb_actions
    if sig==-1 and positions>0:
        cash=cash+(positions*prix_jour)
        positions=0
        
    valeur=cash+positions*prix_jour
    valeurs.append(valeur)
print('valeur finale=',valeurs[-1])
print('cash final =', cash)
print('positions finales =', positions)

pnl = valeurs[-1] - 10000
pnl_pct = (valeurs[-1] / 10000 - 1) * 100

print('PnL =', pnl, 'dh')
print('PnL % =', pnl_pct, '%')
```

```python
# Buy-and-hold : acheter au premier jou
cash_bh=10000
prix_initial=prix.iloc[0]
nbr_actions=int(cash_bh/prix_initial)
cash_bh=cash_bh-nbr_actions*prix_initial
# La valeur du buy-and-hold jour par jour
valeurs_bh=[]
for date in prix.index:
    prix_jour=prix[date]
    valeur_bh=cash_bh+nbr_actions*prix_jour
    valeurs_bh.append(valeur_bh)
    
print("Valeur finale buy-and-hold :", valeurs_bh[-1])
print("Valeur finale ma stratégie :", valeurs[-1])
```
