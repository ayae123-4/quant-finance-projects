```python
connexion = sqlite3.connect(r"C:\Users\HP 1030 G7\Desktop\casa_market.db")
```

```python
def charger_societes(nom, ticker):
    data=load.loadata(nom)
    data.index=pd.to_datetime(data.index,dayfirst=True)
    data=data.reset_index()
    data['ticker']=ticker
    data=data.rename(columns={'Value':'value','Date':'date','Min':'min','Max': 'max', 'Variation': 'variation', 'Volume': 'volume'})
    data.to_sql('cours',connexion, if_exists='append',index=False)
```

```python
charger_societes('Maroc Telecom', 'IAM')
charger_societes('BCP', 'BCP')
charger_societes('LafargeHolcim', 'LHM')
```

```python
connexion.execute('DELETE FROM cours')
connexion.commit()
print('cours videe')
```

```python
import pandas as pd
pd.read_sql('SELECT COUNT(*) FROM cours', connexion)
```
