```python
import numpy as np
import pandas as pd

#Générer des prix aléatoires (marche aléatoire = pur bruit)
np.random.seed(42)                          # pour reproductibilité
rendements_alea = np.random.normal(0, 0.01, 500)   # 500 variations aléatoires
prix_alea = pd.Series(100 * (1 + rendements_alea).cumprod())   # prix qui en découlent

#Calculer le signal SMA sur ces faux prix
sma_alea = prix_alea.rolling(20).mean()
signal_alea = (prix_alea > sma_alea).astype(int).replace(0, -1).shift(1)

#boucle
cash = 10000
positions = 0
valeurs = []

for date in prix_alea.index:
    prix_jour=prix_alea[date]
    sig=signal_alea[date]
    if sig==1 and positions==0:
        nb_actions=int(cash/prix_jour)
        cash=cash-nb_actions*prix_jour
        positions=positions+nb_actions
    if sig==-1 and positions>0:
        cash=cash+(positions*prix_jour)
        positions=0
        
    valeur=cash+positions*prix_jour
    valeurs.append(valeur)

print('valeur finale=',valeurs[-1])
```
