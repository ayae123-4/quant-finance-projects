```python
import yfinance as yf
import numpy as np
import pandas as pd
D=yf.download('AAPL', period='2y')
prix=D['Close'].squeeze()
SMA20=prix.rolling(20).mean()
ecart_type=prix.rolling(20).std()
bande_milieu=SMA20
bande_haut=SMA20+2*ecart_type
bande_bas=SMA20-2*ecart_type 
signal_brut=np.where(prix<bande_bas,1,np.where(prix>bande_haut,-1,0)) #+1 achat, -1 vente, 0 neutre 
signal_brut=pd.Series(signal_brut, index=prix.index)
signal=signal_brut.shift(1)
print(signal_brut.value_counts())
```
