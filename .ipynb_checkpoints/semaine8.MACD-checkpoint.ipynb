```python
import yfinance as yf
import numpy as np 
import pandas as pd
D=yf.download('AAPL', period='2y')
prix=D['Close'].squeeze()
EMA12=prix.ewm(span=12).mean()
EMA26=prix.ewm(span=26).mean()
MACD=EMA12-EMA26
EMA9=MACD.ewm(span=9).mean()
ligne_signal=EMA9
signal_MACD_brut=np.where(MACD>ligne_signal,1,np.where(MACD<ligne_signal,-1,0)) #+1 achat, -1 vente, 0 neutre 
signal_MACD_brut = pd.Series(signal_MACD_brut, index=MACD.index)
signal_MACD=signal_MACD_brut.shift(1)
print(MACD.tail(20))
print(signal_MACD.value_counts()) #ça compte combien de +1, de −1 et de 0 tu obtiens.
```

```python
import yfinance as yf
import numpy as np 
import pandas as pd
D=yf.ticker('AAPL', periode='2y')
prix=D['Close']
EMA(12)=prix.ewm(span=12).mean()
EMA(26)=prix.ewm(span=26).mean()
EMA(9)=prix.ewm(span=9).mean()
MACD=EMA(12)-EMA(26)
ligne_signal=EMA(9)
signal_MACD_brut=np.where(MACD>ligne_signal,1,np.where(MACD<ligne_signal,-1,0)) #+1 achat, -1 vente, 0 neutre 
signal_MACD_brut = pd.Series(signal_MACD_brut, index=MACD.index)
signal_MACD=signal_MACD_brut.shift(1)
print(MACD.tail(20))
print(signal_MACD.value_counts()) #ça compte combien de +1, de −1 et de 0 tu obtiens.
```
