```python
import yfinance as yf
import pandas as pd
import numpy as np
D = yf.download('AAPL', period='2y') 
prix = D['Close'].squeeze() #.squeeze() veut dire si une dimension est inutile (une seule colonne), écrase-la
delta=prix.diff() #1. Variations quotidiennes (prix d'aujourd'hui − prix de la veille)
gains=delta.clip(lower=0) #les variations positives (les négatives deviennent 0)
perte=delta.clip(upper=0).abs() #les variations négatives, en valeur absolue (les positives → 0)
gain_moyen=gains.rolling(14).mean()
perte_moyenne=perte.rolling(14).mean()
RS=gain_moyen/perte_moyenne
RSI=100-(100/(1+RS))
signal_rsi_brut=np.where(RSI<30,1,np.where(RSI>70,-1,0)) #+1 achat, -1 vente, 0 neutre 
signal_rsi_brut = pd.Series(signal_rsi_brut, index=RSI.index)
signal_rsi=signal_rsi_brut.shift(1)
print(RSI.tail(20))
print(signal_rsi.value_counts()) #ça compte combien de +1, de −1 et de 0 tu obtiens.
```
