```python

prix=[485,491,488,495,493,502]
def calcul_rendement(prix_hier,prix_aujourdhui):
    r=(prix_aujourdhui-prix_hier)/prix_hier
    return r
rendement=[]
for i in range (1,len(prix)):
    rd=calcul_rendement(prix[i-1],prix[i])
    rendement.append(rd)
print(rendement)
import numpy as np
mu=np.mean(rendement)
sigma=np.std(rendement, ddof=1)
print(f"moyenne:{mu*100:.2f}%")
print(f"ecart type:{sigma*100:.2f}%")
mu_annuel=mu*252
sigma_annuel=sigma*np.sqrt(252)
sharpe=mu/sigma
sharpe_annuel=sharpe*np.sqrt(252)
print(mu_annuel*100,"%",sigma_annuel*100,"%",sharpe,sharpe_annuel)
```

```python
import yfinance as yf
import matplotlib.pyplot as plt

apple = yf.Ticker("AAPL")
dr = apple.history(period="2y")
df = apple.history(period="3mo")
ds = apple.history(period="1y")
dt = apple.history(period="2y")

# Calculer la moyenne mobile sur 20 jours
df["SMA_20"] = df["Close"].rolling(20).mean()
ds["SMA_50"] = ds["Close"].rolling(50).mean()
dt["SMA_200"] = dt["Close"].rolling(200).mean()

# Tracer les deux courbes
plt.plot(dr["Close"], label="Prix de clôture", color="blue")
plt.plot(df["SMA_20"], label="Moyenne mobile 20j", color="orange")
plt.plot(ds["SMA_50"], label="Moyenne mobile 50j", color="red")
plt.plot(dt["SMA_200"], label="Moyenne mobile 200j", color="black")

plt.title("Apple - Prix et Moyenne Mobile")
plt.xlabel("Date")
plt.ylabel("Prix en $")
plt.legend()
plt.show()
```

```python
import yfinance as yf
import matplotlib.pyplot as plt

# Télécharger les données
apple = yf.Ticker("AAPL")
df = apple.history(period="6mo")

# Calculer les SMA
df["SMA_20"] = df["Close"].rolling(20).mean()
df["SMA_50"] = df["Close"].rolling(50).mean()

# Tracer le graphique
plt.figure(figsize=(12, 6))

plt.plot(df["Close"],  label="Prix de clôture", color="blue",   linewidth=1.5)
plt.plot(df["SMA_20"], label="SMA 20",          color="orange", linewidth=1.5)
plt.plot(df["SMA_50"], label="SMA 50",          color="red",    linewidth=1.5)

plt.title("Apple — Prix et Moyennes Mobiles")
plt.xlabel("Date")
plt.ylabel("Prix en $")
plt.legend()
plt.grid(True)
plt.show()
```
